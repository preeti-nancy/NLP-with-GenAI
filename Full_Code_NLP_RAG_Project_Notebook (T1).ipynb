## Question Answering using LLM

#### Downloading and Loading the model

```python
# Download the Mistral-7B-Instruct model from Hugging Face Hub
model_name = "TheBloke/Mistral-7B-Instruct-v0.1-GGUF"
model_file = "mistral-7b-instruct-v0.1.Q4_K_M.gguf"

# Download the model
model_path = hf_hub_download(
    repo_id=model_name,
    filename=model_file,
    repo_type="model"
)

print(f"Model downloaded at: {model_path}")

# Load the model using llama-cpp-python
llm = Llama(
    model_path=model_path,
    n_ctx=4096,  # Context window
    n_threads=8,  # Number of CPU threads to use
    n_gpu_layers=35  # Number of layers to offload to GPU (set to 0 if using CPU only)
)

print("Model loaded successfully!")
```

#### Response

```python
def response(query,max_tokens=128,temperature=0,top_p=0.95,top_k=50):
    model_output = llm(
      prompt=query,
      max_tokens=max_tokens,
      temperature=temperature,
      top_p=top_p,
      top_k=top_k
    )

    return model_output['choices'][0]['text']
```

## Question Answering using LLM with Prompt Engineering

### Query 1: What is the protocol for managing sepsis in a critical care unit?

```python
query1 = "What is the protocol for managing sepsis in a critical care unit?"

print("="*80)
print("COMBINATION 1: Zero-shot with Medical Expert Role")
print("="*80)
prompt1 = f"""You are an experienced critical care physician. {query1}
Provide a detailed, step-by-step protocol."""

resp1 = response(prompt1, max_tokens=300, temperature=0, top_p=0.95, top_k=50)
print(resp1)
print("\nObservation: Adding role-based context helps guide the model to provide structured medical protocols.\n")

print("="*80)
print("COMBINATION 2: Few-shot with Example")
print("="*80)
prompt2 = f"""Answer medical questions with detailed protocols.

Example:
Q: What is the protocol for cardiac arrest?
A: 1. Call for help, 2. Start CPR, 3. Use AED, 4. Administer medications...

Q: {query1}
A:"""

resp2 = response(prompt2, max_tokens=300, temperature=0.3, top_p=0.9, top_k=40)
print(resp2)
print("\nObservation: Few-shot learning with examples improves structured response format. Temperature=0.3 adds slight variation.\n")

print("="*80)
print("COMBINATION 3: Chain-of-Thought Reasoning")
print("="*80)
prompt3 = f"""{query1}
Let's think step by step about the critical care protocol for sepsis management, including recognition, resuscitation, and treatment phases."""

resp3 = response(prompt3, max_tokens=350, temperature=0.5, top_p=0.85, top_k=30)
print(resp3)
print("\nObservation: CoT prompting encourages logical reasoning. Temperature=0.5 allows more creative but still controlled responses.\n")

print("="*80)
print("COMBINATION 4: Structured Output Format")
print("="*80)
prompt4 = f"""As a medical AI assistant, answer the following question in this format:
- Initial Assessment:
- Immediate Interventions:
- Ongoing Management:
- Monitoring:

Question: {query1}"""

resp4 = response(prompt4, max_tokens=400, temperature=0.1, top_p=0.95, top_k=50)
print(resp4)
print("\nObservation: Explicit formatting instructions create organized, scannable responses. Low temperature maintains precision.\n")

print("="*80)
print("COMBINATION 5: High Temperature for Comprehensive Coverage")
print("="*80)
prompt5 = f"""You are a medical knowledge base. Provide comprehensive information about: {query1}
Include all relevant details about identification, treatment, and monitoring."""

resp5 = response(prompt5, max_tokens=400, temperature=0.7, top_p=0.9, top_k=60)
print(resp5)
print("\nObservation: Higher temperature (0.7) with increased tokens explores broader response space but may reduce precision.\n")

print("="*80)
print("SUMMARY OF OBSERVATIONS:")
print("="*80)
print("""
1. Role-based prompts (expert persona) improve domain-specific responses
2. Few-shot examples help establish desired output structure
3. Chain-of-thought prompting enhances logical progression
4. Explicit formatting creates more usable, organized outputs
5. Temperature tuning: 0.0-0.1 (precise), 0.3-0.5 (balanced), 0.7+ (creative but riskier)
6. Top_p and top_k control diversity; lower values = more focused
7. Prompt engineering significantly impacts response quality even without retrieval
""")
```

## Data Preparation for RAG

### Loading the Data

```python
# Load the Merck Manual PDF
pdf_path = "Merck_Manual.pdf"

loader = PyMuPDFLoader(pdf_path)
documents = loader.load()

print(f"Successfully loaded {len(documents)} pages from the Merck Manual")
print(f"Document type: {type(documents)}")
print(f"First document type: {type(documents[0])}")
```

### Data Overview

#### Checking the first 5 pages

```python
# Display first 5 pages content
for i in range(min(5, len(documents))):
    print(f"\n{'='*80}")
    print(f"PAGE {i+1}")
    print(f"{'='*80}")
    print(f"Page Content Length: {len(documents[i].page_content)} characters")
    print(f"\nFirst 500 characters:")
    print(documents[i].page_content[:500])
    print(f"\nMetadata: {documents[i].metadata}")
```

#### Checking the number of pages

```python
total_pages = len(documents)
print(f"Total number of pages in the Merck Manual: {total_pages}")

# Calculate total characters
total_chars = sum(len(doc.page_content) for doc in documents)
print(f"Total characters across all pages: {total_chars:,}")
print(f"Average characters per page: {total_chars/total_pages:.2f}")
```

### Data Chunking

```python
# Split documents into chunks
# Using RecursiveCharacterTextSplitter for intelligent chunking
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,        # Size of each chunk in characters
    chunk_overlap=200,      # Overlap between chunks to maintain context
    length_function=len,    # Function to measure chunk length
    separators=["\n\n", "\n", " ", ""]  # Priority order for splitting
)

chunks = text_splitter.split_documents(documents)

print(f"Total number of chunks created: {len(chunks)}")
print(f"Original pages: {len(documents)}")
print(f"Chunks per page (average): {len(chunks)/len(documents):.2f}")

# Display sample chunks
print("\n" + "="*80)
print("SAMPLE CHUNK 1:")
print("="*80)
print(chunks[0].page_content[:300])
print(f"\nChunk metadata: {chunks[0].metadata}")

print("\n" + "="*80)
print("SAMPLE CHUNK 2:")
print("="*80)
print(chunks[100].page_content[:300])
print(f"\nChunk metadata: {chunks[100].metadata}")
```

### Embedding

```python
# Create embedding function using SentenceTransformers
# Using a medical/scientific domain model for better performance
embedding_function = SentenceTransformerEmbeddings(
    model_name="all-MiniLM-L6-v2"  # Fast and efficient model
)

print("Embedding model loaded successfully!")
print(f"Model name: all-MiniLM-L6-v2")

# Test the embedding function
sample_text = "What are the symptoms of sepsis?"
sample_embedding = embedding_function.embed_query(sample_text)

print(f"\nSample embedding dimensions: {len(sample_embedding)}")
print(f"First 10 values of sample embedding: {sample_embedding[:10]}")
```

### Vector Database

```python
# Create Chroma vector database from chunks
vectordb = Chroma.from_documents(
    documents=chunks,
    embedding=embedding_function,
    persist_directory="./chroma_db",  # Directory to persist the database
    collection_name="merck_manual"
)

print("Vector database created successfully!")
print(f"Collection name: merck_manual")
print(f"Number of documents in database: {vectordb._collection.count()}")

# Test similarity search
test_query = "sepsis treatment"
test_results = vectordb.similarity_search(test_query, k=3)

print(f"\nTest similarity search for: '{test_query}'")
print(f"Number of results: {len(test_results)}")
print("\nFirst result preview:")
print(test_results[0].page_content[:200])
```

### Retriever

```python
# Create retriever from vector database
retriever = vectordb.as_retriever(
    search_type="similarity",  # Use similarity search
    search_kwargs={"k": 3}     # Retrieve top 3 most relevant chunks
)

print("Retriever created successfully!")
print(f"Search type: similarity")
print(f"Default k (number of chunks to retrieve): 3")

# Test the retriever
test_query = "What are the symptoms of appendicitis?"
retrieved_docs = retriever.get_relevant_documents(test_query)

print(f"\nTest retrieval for query: '{test_query}'")
print(f"Number of documents retrieved: {len(retrieved_docs)}")

for i, doc in enumerate(retrieved_docs, 1):
    print(f"\n--- Retrieved Document {i} ---")
    print(f"Content preview: {doc.page_content[:200]}...")
    print(f"Source page: {doc.metadata.get('page', 'Unknown')}")
```

```python
# Define system message for the RAG QA system
qna_system_message = """You are a medical AI assistant with access to the Merck Manual.
Your role is to provide accurate, evidence-based medical information based on the context provided.
Always base your answers on the given context from the medical manual.
If the context doesn't contain enough information to fully answer the question, acknowledge this limitation.
Provide clear, structured, and professional medical information."""

# Define user message template
qna_user_message_template = """Context from Merck Manual:
{context}

Question: {question}

Based on the context provided above, please provide a comprehensive and accurate answer:"""

print("System and User Prompt Templates Created Successfully!")
print("\n" + "="*80)
print("SYSTEM MESSAGE:")
print("="*80)
print(qna_system_message)
print("\n" + "="*80)
print("USER MESSAGE TEMPLATE:")
print("="*80)
print(qna_user_message_template)
```

### Response Function

```python
def generate_rag_response(user_input,k=3,max_tokens=128,temperature=0,top_p=0.95,top_k=50):
    global qna_system_message,qna_user_message_template
    # Retrieve relevant document chunks
    relevant_document_chunks = retriever.get_relevant_documents(query=user_input,k=k)
    context_list = [d.page_content for d in relevant_document_chunks]

    # Combine document chunks into a single context
    context_for_query = ". ".join(context_list)

    user_message = qna_user_message_template.replace('{context}', context_for_query)
    user_message = user_message.replace('{question}', user_input)

    prompt = qna_system_message + '\n' + user_message

    # Generate the response
    try:
        response = llm(
                  prompt=prompt,
                  max_tokens=max_tokens,
                  temperature=temperature,
                  top_p=top_p,
                  top_k=top_k
                  )

        # Extract and print the model's response
        response = response['choices'][0]['text'].strip()
    except Exception as e:
        response = f'Sorry, I encountered the following error: \n {e}'

    return response
```

## Output Evaluation

Let us now use the LLM-as-a-judge method to check the quality of the RAG system on two parameters - retrieval and generation. We illustrate this evaluation based on the answeres generated to the question from the previous section.

- We are using the same Mistral model for evaluation, so basically here the llm is rating itself on how well he has performed in the task.

```python
groundedness_rater_system_message = """You are an expert evaluator assessing the groundedness of AI-generated answers.

Groundedness measures whether the answer is fully supported by and derived from the provided context.

Evaluate the answer based on these criteria:
1. All factual claims in the answer must be present in the context
2. The answer should not introduce information not found in the context
3. The answer should not contradict the context
4. Inferences must be logical extensions of the context

Provide a score from 1-5:
5 - Completely grounded: All information directly from context, no unsupported claims
4 - Mostly grounded: Minor unsupported details but core information is from context
3 - Partially grounded: Some information from context, some from external knowledge
2 - Minimally grounded: Mostly external information, little from context
1 - Not grounded: Answer ignores context or contradicts it

Respond with the score and a brief justification."""
```

```python
relevance_rater_system_message = """You are an expert evaluator assessing the relevance of AI-generated answers to user questions.

Relevance measures how well the answer addresses the specific question asked.

Evaluate the answer based on these criteria:
1. Does the answer directly address the question?
2. Does the answer cover all aspects mentioned in the question?
3. Is the information provided appropriate to the question's scope?
4. Does the answer stay focused on the question without excessive tangents?

Provide a score from 1-5:
5 - Highly relevant: Directly and completely addresses all aspects of the question
4 - Mostly relevant: Addresses main aspects with minor gaps or tangents
3 - Moderately relevant: Addresses some aspects but misses key points or includes irrelevant info
2 - Minimally relevant: Tangentially related but doesn't address core question
1 - Not relevant: Answer doesn't address the question

Respond with the score and a brief justification."""
```

```python
user_message_template = """Context: {context}

Question: {question}

Answer: {answer}

Please evaluate the above answer based on the criteria provided."""
```

```python
def generate_ground_relevance_response(user_input,k=3,max_tokens=128,temperature=0,top_p=0.95,top_k=50):
    global qna_system_message,qna_user_message_template
    # Retrieve relevant document chunks
    relevant_document_chunks = retriever.get_relevant_documents(query=user_input,k=3)
    context_list = [d.page_content for d in relevant_document_chunks]
    context_for_query = ". ".join(context_list)

    # Combine user_prompt and system_message to create the prompt
    prompt = f"""[INST]{qna_system_message}\n
                {'user'}: {qna_user_message_template.format(context=context_for_query, question=user_input)}
                [/INST]"""

    response = llm(
            prompt=prompt,
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
            top_k=top_k,
            stop=['INST'],
            )

    answer =  response["choices"][0]["text"]

    # Combine user_prompt and system_message to create the prompt
    groundedness_prompt = f"""[INST]{groundedness_rater_system_message}\n
                {'user'}: {user_message_template.format(context=context_for_query, question=user_input, answer=answer)}
                [/INST]"""

    # Combine user_prompt and system_message to create the prompt
    relevance_prompt = f"""[INST]{relevance_rater_system_message}\n
                {'user'}: {user_message_template.format(context=context_for_query, question=user_input, answer=answer)}
                [/INST]"""

    response_1 = llm(
            prompt=groundedness_prompt,
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
            top_k=top_k,
            stop=['INST'],
            )

    response_2 = llm(
            prompt=relevance_prompt,
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
            top_k=top_k,
            stop=['INST'],
            )

    return response_1['choices'][0]['text'],response_2['choices'][0]['text']
```
